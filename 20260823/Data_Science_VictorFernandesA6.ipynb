9. Crie um array com 10 números aleatórios e normalize seus valores entre 0

```python
import numpy as np

# Entendendo que nessa questão, deve-se pegar um array de 10 valores aleatórios
# e transforma-los em números entre 0 e 1, sendo 1 o maior número do array e 0
# o menor:

# cria array com 10 números aleatórios
lista = np.random.randint(1,100,10)
print(f"Antes: {lista}")

# cálculo de normalização: x_n = (x-min_x)/(max_x-min_x)
maior_lista = np.max(lista)
menor_lista = np.min(lista)

lista = (lista-menor_lista)/(maior_lista-menor_lista)

print(f"depois {lista}")
```

19. Crie dois arrays de números inteiros aleatórios de tamanho 20 e calcule o
ângulo entre eles (produto escalar normalizado)

```python
import numpy as np

# cria dois arrays de tamanho 20
a = np.random.randint(1,50,20)
b = np.random.randint(1,50,20)

# Calculo: a*b/||a||*||b||

# a*b
ab = np.sum(a*b)

# ||a||
comprimento_a = np.sqrt(np.sum(a*a))

# ||b||
comprimento_b = np.sqrt(np.sum(b*b))

# retorna ângulo
angulo = ab/(comprimento_a*comprimento_b)

print(a)
print(b)
print(f"angulo entre a e b (em radianos): {angulo}")
```

4. Use a função de otimização de SciPy para maximizar f(x) = -x^2 + 4x + 1

```python
from scipy import optimize

# funcao f(x)
def f(x):
  return -(x**2) + 4*x + 1

# para maximizar a função, é preciso minimizar o oposto da função
def oposto_f(x):
  return -f(x)

resultado = optimize.minimize(oposto_f,1)

print(f"funcao f(x) = -x^2 + 4x + 1 maximizada\nx = {resultado.x[0]:.3f}\nmaximo = {resultado.fun}")
```

9. Ajuste uma curva quadrática aos pontos (1, 2),(2, 4),(3, 5),(4, 7),(5, 8)

```python
from scipy import optimize
from numpy import array

# array dos pontos
pontos_x = array([1,2,3,4,5])
pontos_y = array([2,4,5,7,8])

# funcao da curva quadrática
def f(x,a,b,c):
  return a*x**2 + b*x + c

# retorna os parâmetros de ajuste
parametros, _ = optimize.curve_fit(f,pontos_x,pontos_y)

print("Parametros (a,b,c) ajustados para a curva quadratica:\n",parametros)
```

16.  Resolva a equação diferencial de segunda ordem d^2y/dx^2 + y = 0, com condições iniciais y(0) = 1 e y' (0) = 0

```python
from scipy import integrate

# funcao
def f(x,y):
  # y1 e y2
  y1, y2 = y

  #dy1/dx e dy2/dx
  dy1_dx = y2
  dy2_dx = -y1

  return[dy1_dx,dy2_dx]

# array com as condições iniciais
condicoes = [1,0]

resultado = integrate.solve_ivp(f,[0,10],condicoes)

print("Resultado: ",resultado.y[0])
```

```python
from scipy import linalg
from numpy import array

# exemplo de sistema de equações lineares com tres variaveis:
#
# x = 10; y = 3; z = 2
# 2*x - 4*y  + 2*z  = 12
# x   + y    + z    = 15
# 5*x - 10*y - 10*z = 0

# array com os multiplicadores das variaveis
a = array([[2,-4,2],[1,1,1],[5,-10,-10]])

# array com as respostas das variaveis
b = array([12,15,0])

resultado = linalg.solve(a,b)

print("sistema de equacoes:\n")
print("2*x\t-\t4*y\t+\t2*z\t=\t12\nx\t+\ty\t+\tz\t=\t15\n5*x\t-\t10*y\t-\t10*z\t=\t0\n")
print("Valores de [x,y,z]:",resultado)

```

1. Crie um gráfico de linha para a função y = x3 no intervalo de x = -10 até x = 10

2. Plote um gráfico de dispersão para 100 pontos aleatórios com uma paleta de cores baseada nos valores de y

```python
import matplotlib.pyplot as plt
import matplotlib.colors as matc
import numpy as np

# Dados X e Y
x = np.random.randint(0,100,100)
y = np.random.randint(0,100,100)

# cores
#matc.Normalize() -> cria uma normalização com base nos valores mínimo e máximo
cores = matc.Normalize(np.min(y),np.max(y),False)(y)

# gráfico
plt.scatter(x,y,c = cores)
plt.title("Dispersão de pontos aleatorios")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```

a função y = sin(x)

```python
import matplotlib.pyplot as plt
import numpy as np

# dados
x = np.linspace(0,10,100)
y = np.sin(x)

# intervalo de confiança
ic = 0.2

plt.plot(x,y)
plt.fill_between(x,y-ic,y+ic, color = "blue", alpha = 0.2)
plt.title("Grafico de linhas com intervalo de confianca sombreado")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
